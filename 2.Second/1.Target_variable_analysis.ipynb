I want to detect those and remove because i want to train my model to understand one entry as only one vehicle .

```python
bulk_keywords = (
    r'fleet sale|entire fleet|bulk sale|'
    r'multiple vehicles|multiple cars|'
    r'multiple trucks|lot of vehicles|'
    r'package deal|selling all|'
    r'collection of cars|collection of trucks'
)

bulk_mask = df['description'].str.contains(
    bulk_keywords,
    case=False,
    na=False,
    regex=True
)

print("Potential bulk listings:", bulk_mask.sum())
```

```python
df.loc[
    bulk_mask,
    ['price', 'year', 'manufacturer', 'model', 'title', 'description']
].head(15)
```

### Target Variable (price) – Data Quality Investigation

1. Zero-priced listings**
- Approximately 10,000 listings (3.4%) have a recorded price of $0.
- Some descriptions contain actual selling prices, but others mention monthly payments, down payments, or undisclosed prices.
- These records will be excluded from model training for now because their actual selling prices cannot be reliably determined.

2. Extremely high-priced listings**
- Several listings contain unrealistic prices, including values above $1 million and even billions.
- Some appear to be placeholder prices, incorrect entries, wanted ads, or bulk vehicle advertisements.
- However, some expensive older vehicles may be genuine collector or classic cars.
- Therefore, we should investigate extreme values rather than automatically remove all expensive or older vehicles.

Conclusion: The target variable contains zero values, extreme outliers, and potentially incorrect entries. Further cleaning and validation will be necessary before model training.
